***3.17. Môi trường tính toán Graph trong tensorflow***

```python
def a_regular_function(x, y, b):
  x = tf.matmul(x, y)
  x = x + b
  return x
#'a_function_that_uses_a_graph' la 1 ham TensorFlow
a_function_that_uses_a_graph = tf.function(a_regular_function)

# tao vai tensors
x1= tf.constant([[1.0, 2.0]])
y1 = tf.constant([[2.0], [3.0]])
b1 = tf.constant(4.0)

orig_value = a_regular_function(x1, y1, b1).numpy()
# Goi ham
tf_function_value = a_function_that_uses_a_graph(x1, y1, b1).numpy()
print(orig_value)
print(tf_function_value)
```

*3.17.1.Biến một hàm Python thuần túy sang graph*

```python
def inner_function(x, y, b):
  x = tf.matmul(x, y)
  x = x + b
  return x

#
@tf.function
def outer_function(x):
  y = tf.constant([[2.0], [3.0]])
  b = tf.constant(4.0)

  return inner_function(x, y, b)

#
#
rs = outer_function(tf.constant([[1.0, 2.0]])).numpy()
print(rs)
```

***3.18. Modules, Layers và Models***

```python
class Dense(tf.keras.layers.Layer):
  def __init__(self, in_features, out_features , name=None):
    super().__init__(name=name)
    self.w = tf.Variable(
        tf.random.normal([in_features, out_features]), name='w')
    self.b = tf.Variable(tf.zeros([out_features]), name='b')
  def call(self, x):
    y = tf.matmul(x, self.w) + self.b
    return tf.nn.relu(y)

class SequentialModule(tf.Module):
  def __init__(self, name=None):
    super().__init__(name=name)

    self.dense_1 = Dense(in_features=3, out_features=3)
    self.dense_2 = Dense(in_features=3, out_features=2)

  def __call__(self, x):
    x = self.dense_1(x)
    return self.dense_2(x)

#
my_model = SequentialModule(name="the_model")

#
print("Model results:", my_model(tf.constant([[2.0, 2.0, 2.0]])))
```

***3.19. Lưu weight***

```python
#3.19.1. Lưu checkpoint
chkp_path = "my_checkpoint"
checkpoint = tf.train.Checkpoint(model=my_model)
checkpoint.write(chkp_path)
```

```python
#3.19.2. Phục hồi checkpoint
new_model = SequentialModule()
new_checkpoint = tf.train.Checkpoint(model=new_model)
new_checkpoint.restore("my_checkpoint")

print(new_model(tf.constant([[2.0, 2.0, 2.0]])))
```
